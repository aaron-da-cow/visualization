```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
tips=sns.load_dataset('tips')
print(tips)
```

```python
sns.lineplot(data=tips,x='total_bill',y='tip')
```

```python
sns.set_style('dark')
sns.lineplot(data=tips,x='total_bill',y='tip',
             color='hotpink',
             linestyle='--',
             linewidth=3,
             marker='o',
             markerfacecolor='indigo')
tipgraph=sns.lineplot(data=tips, x='total_bill',y='tip',
                      color='hotpink',
                      linestyle='--',
                      linewidth=3,
                      marker='o',
                      markerfacecolor='indigo')
tipgraph.set(title="Tips vs. Total Bill",
             xlabel="Total Bill ($)",
             ylabel="Tip Amount ($)")
```

```python
tipgraph=sns.scatterplot(data=tips,x='total_bill',y='tip',style='time', hue='day', palette = ['purple', 'hotpink', 'deepskyblue', 'yellowgreen'])
tipgraph.set(title='Tips vs Total Bill',
             xlabel='Total Bill ($)',
             ylabel='Tip Amount ($)')
```

```python
#sns.pairplot(data=tips)
sns.pairplot(data = tips, hue = 'day')
```

```python
daysplot = sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    col="day",
    kind="scatter",
    col_wrap=2
)
```

```python
import plotly.graph_objects as go
x1=np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110,180,240,99,220])
```

```python
graph = go.Figure()
graph.add_trace(go.Bar(x=x1,y=y1))
graph.update_layout(title='Pirate_Scores',
                    xaxis_title="Pirates",
                    yaxis_title="Score")
```

Wordclouds and Venn Diagrams

```python
from wordcloud import WordCloud
df = pd.read_csv("https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv",on_bad_lines='skip')
print(df)
```

```python
text = " ".join(each for each in df.quote)
wordcloud = WordCloud(background_color="white",
                      colormap= 'inferno').generate(text)
fig, ax = plt.subplots(figsize=(7,3))
ax.imshow(wordcloud, interpolation='bilinear')
ax.axis("off")
```

```python
from matplotlib_venn import venn2, venn2_circles, venn2_unweighted
A = set(["apple", "banana", "watermelon"])
B = set(["pumpkin", "blueberry", "apple", "key lime"])
```

```python
diagram=venn2_unweighted([A,B], set_labels=('Fruits', 'Pies'),
                         set_colors=("blue", "red"),
                         alpha=0.5)
diagram.get_label_by_id("10").set_text("\n".join(A-B))
diagram.get_label_by_id("11").set_text('\n'.join(A&B))
diagram.get_label_by_id("01").set_text('\n'.join(B-A))
```

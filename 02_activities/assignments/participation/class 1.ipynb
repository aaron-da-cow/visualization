Visualizations should be pleasing to look at, accurately present data and conveys a message. 


```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import PIL
import requests
```

```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0, 100,50)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y)
```

```python
fig,ax=plt.subplots(figsize=(5,3))
ax.plot(x,y)
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
fig.tight_layout()
```

```python
fig,ax=plt.subplots(figsize=(5,3))
ax.scatter(x,y,marker='*',color="indigo")
plt.show()
```

```python
fig,ax=plt.subplots(figsize=(5,3))
ax.plot(x,y,marker='*',color='indigo',linestyle='--',linewidth=2)
```

```python
# libraries
import numpy as np
import matplotlib.pyplot as plt
 
# create dataset
height = [3, 12, 5, 18, 45]
bars = ('A', 'B', 'C', 'D', 'E')
x_pos = np.arange(len(bars))
 
# Create bars and choose color
plt.bar(x_pos, height, color = (0.5,0.1,0.5,0.6))
 
# Add title and axis names
plt.title('My title')
plt.xlabel('categories')
plt.ylabel('values')
 
# Create names on the x axis
plt.xticks(x_pos, bars)
 
# Show graph
plt.show()
```

It's clear and substantive, but probably not aesthetically pleasing.

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import PIL
import requests
#all the required libraries
```

```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0, 100,50)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
#scatterplot code
ax.scatter(x,y)
```

```python
fig,ax=plt.subplots(figsize=(5,3))
ax.plot(x,y)
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
fig.tight_layout()
```

Adding comments helps the intelligibility of the code.

```python
y1 = np.random.randint(0, 100,50)
y2 = np.random.randint(0, 100,50)

fig,ax=plt.subplots(figsize=(5,3))
ax.plot(x,y1)
ax.plot(x,y2)
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
fig.tight_layout()
```
